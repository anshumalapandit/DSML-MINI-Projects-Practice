```python
# ----------------------------------------------
# STEP 1: Store all the given points
# ----------------------------------------------
P1 = (0.1, 0.6)
P2 = (0.15, 0.71)
P3 = (0.08, 0.9)
P4 = (0.16, 0.85)
P5 = (0.2, 0.3)
P6 = (0.25, 0.5)
P7 = (0.24, 0.1)
P8 = (0.3, 0.2)

all_points = [P1, P2, P3, P4, P5, P6, P7, P8]

# ----------------------------------------------
# STEP 2: Initialize centroids
# m1 = P1 and m2 = P8
# ----------------------------------------------
m1 = P1   # centroid of Cluster 1
m2 = P8   # centroid of Cluster 2

# ----------------------------------------------
# STEP 3: Define easy distance formula
# Using Euclidean distance
# ----------------------------------------------
def distance(p, m):
    # sqrt( (x1-x2)^2 + (y1-y2)^2 )
    return ((p[0] - m[0])**2 + (p[1] - m[1])**2)**0.5

# ----------------------------------------------
# STEP 4: Assign each point to the nearest centroid
# ----------------------------------------------
C1 = []   # Cluster 1 points
C2 = []   # Cluster 2 points

for p in all_points:
    d1 = distance(p, m1)
    d2 = distance(p, m2)

    # Assign to whichever centroid is closer
    if d1 < d2:
        C1.append(p)
    else:
        C2.append(p)

# ----------------------------------------------
# STEP 5: Compute new centroid m1 (average of C1)
# ----------------------------------------------
x1 = sum(p[0] for p in C1) / len(C1)
y1 = sum(p[1] for p in C1) / len(C1)
m1_new = (x1, y1)

# ----------------------------------------------
# STEP 6: Compute new centroid m2 (average of C2)
# ----------------------------------------------
x2 = sum(p[0] for p in C2) / len(C2)
y2 = sum(p[1] for p in C2) / len(C2)
m2_new = (x2, y2)

# ----------------------------------------------
# STEP 7: Print final answers
# ----------------------------------------------
print("Cluster 1:", C1)
print("Cluster 2:", C2)

# 1) Which cluster does P6 belong to?
print("P6 belongs to Cluster:", "C1" if P6 in C1 else "C2")

# 2) What is the population of Cluster around m2?
print("Population of cluster around m2:", len(C2))

# 3) What are the updated centroids?
print("Updated m1:", m1_new)
print("Updated m2:", m2_new)
```

17. Compute Accuracy, Error rate, Precision, Recall for following confusion
matrix ( Use formula for each)

True Positives (TPs): 1 False Positives (FPs): 1
False Negatives (FNs): 8 True Negatives (TNs): 90

```python
# Given confusion matrix values
TP = 1   # True Positives
FP = 1   # False Positives
FN = 8   # False Negatives
TN = 90  # True Negatives

# Total samples
total = TP + FP + FN + TN

# Accuracy
accuracy = (TP + TN) / total

# Error Rate
error_rate = (FP + FN) / total

# Precision
precision = TP / (TP + FP)

# Recall
recall = TP / (TP + FN)

# Print results
print("Accuracy:", accuracy)
print("Error Rate:", error_rate)
print("Precision:", precision)
print("Recall:", recall)
```

20. Write a program to cluster a set of points using K-means for IRIS
dataset. Consider, K=3, clusters. Consider Euclidean distance as the
distance measure. Randomly initialize a cluster mean as one of the data
points. Iterate at least for 10 iterations. After iterations are over, print the
final cluster means for each of the clusters.

```python
import pandas as pd
import random
import math

# Load the data (use the uploaded file)
df = pd.read_csv("/content/IRIS.csv")

# Use only numeric columns for clustering
data = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].values.tolist()

K = 3
random.seed(42)

# Pick 3 random initial centroids (choose 3 points from data)
indices = random.sample(range(len(data)), K)
centroids = [data[i] for i in indices]

# Run 10 iterations
for it in range(10):
    # create empty clusters
    clusters = [[] for _ in range(K)]

    # Assignment step: assign each point to nearest centroid
    for point in data:
        # compute distances to each centroid
        d0 = math.sqrt((point[0]-centroids[0][0])**2 + (point[1]-centroids[0][1])**2 + (point[2]-centroids[0][2])**2 + (point[3]-centroids[0][3])**2)
        d1 = math.sqrt((point[0]-centroids[1][0])**2 + (point[1]-centroids[1][1])**2 + (point[2]-centroids[1][2])**2 + (point[3]-centroids[1][3])**2)
        d2 = math.sqrt((point[0]-centroids[2][0])**2 + (point[1]-centroids[2][1])**2 + (point[2]-centroids[2][2])**2 + (point[3]-centroids[2][3])**2)

        # find smallest distance and append to that cluster
        if d0 <= d1 and d0 <= d2:
            clusters[0].append(point)
        elif d1 <= d0 and d1 <= d2:
            clusters[1].append(point)
        else:
            clusters[2].append(point)

    # Update step: compute new centroids (mean of each cluster)
    new_centroids = []
    for c in clusters:
        if len(c) == 0:
            new_centroids.append(centroids[len(new_centroids)])  # keep old if empty
        else:
            sx = sum(p[0] for p in c)
            sy = sum(p[1] for p in c)
            sz = sum(p[2] for p in c)
            sw = sum(p[3] for p in c)
            n = len(c)
            new_centroids.append([sx/n, sy/n, sz/n, sw/n])
    centroids = new_centroids
    print(f"Iteration {it+1}: cluster sizes = {[len(c) for c in clusters]}")

# Final centroids
print("\nFinal centroids:")
for i, c in enumerate(centroids, 1):
    print(f" m{i} =", [round(v,4) for v in c])
```

### **Assignment 21**

21. Write a program to cluster a set of points using K-means for IRIS
dataset. Consider, K=4, clusters. Consider Euclidean distance as the
distance measure. Randomly initialize a cluster mean as one of the data
points. Iterate at least for 10 iterations. After iterations are over, print the
final cluster means for each of the clusters.

```python
import pandas as pd
import random
import math

# Load dataset
df = pd.read_csv("/content/IRIS.csv")

# Only numeric features
data = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].values.tolist()

# K = 4 clusters
K = 4
random.seed(42)

# Randomly choose 4 initial centroids
initial_indices = random.sample(range(len(data)), K)
centroids = [data[i] for i in initial_indices]

print("Initial centroids:")
for i, c in enumerate(centroids, 1):
    print(f"m{i} =", c)

# Function for Euclidean distance
def distance(p, c):
    return math.sqrt(
        (p[0]-c[0])**2 +
        (p[1]-c[1])**2 +
        (p[2]-c[2])**2 +
        (p[3]-c[3])**2
    )

# Run K-means for 10 iterations
for it in range(10):
    clusters = [[] for _ in range(K)]

    # Assign points to nearest centroid
    for point in data:
        d_list = [distance(point, centroids[i]) for i in range(K)]
        nearest = d_list.index(min(d_list))
        clusters[nearest].append(point)

    # Update centroids = mean of cluster points
    new_centroids = []
    for cluster in clusters:
        if len(cluster) == 0:
            new_centroids.append([0,0,0,0])  # avoid division error
        else:
            x = sum(p[0] for p in cluster)/len(cluster)
            y = sum(p[1] for p in cluster)/len(cluster)
            z = sum(p[2] for p in cluster)/len(cluster)
            w = sum(p[3] for p in cluster)/len(cluster)
            new_centroids.append([x, y, z, w])

    centroids = new_centroids
    print(f"Iteration {it+1}: cluster sizes = {[len(c) for c in clusters]}")

# Print final 4 centroids
print("\nFinal centroids after 10 iterations:")
for i, c in enumerate(centroids, 1):
    print(f"Cluster {i} centroid:", [round(val,4) for val in c])
```

### **Assignment 23**

23. With reference to Table , obtain the Frequency table for the
attribute age. From the frequency table you have obtained, calculate the
information gain of the frequency table while splitting on Age. (Use step
by step Python/Pandas commands)

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_excel("/content/Age Table.xlsx")

# STEP 1: Frequency table
freq = pd.crosstab(df["Age"], df["Class"])
print("Frequency Table:\n", freq, "\n")

# Entropy function
def entropy(col):
    probs = col.value_counts(normalize=True)
    return -(probs * np.log2(probs)).sum()

# STEP 2: Parent entropy
entropy_full = entropy(df["Class"])
print("Entropy of Full Dataset:", entropy_full, "\n")

# STEP 3: Entropy of each Age group
entropy_young = entropy(df[df["Age"]=="Young"]["Class"])
entropy_middle = entropy(df[df["Age"]=="Middle"]["Class"])
entropy_old = entropy(df[df["Age"]=="Old"]["Class"])

print("Entropy Young:", entropy_young)
print("Entropy Middle:", entropy_middle)
print("Entropy Old:", entropy_old, "\n")

# STEP 4: Weighted entropy after splitting on Age
N = len(df)
entropy_after = (
    (len(df[df["Age"]=="Young"]) / N) * entropy_young +
    (len(df[df["Age"]=="Middle"]) / N) * entropy_middle +
    (len(df[df["Age"]=="Old"]) / N) * entropy_old
)

print("Entropy After Split:", entropy_after, "\n")

# STEP 5: Information Gain
IG = entropy_full - entropy_after
print("Information Gain for Age:", IG)
```

**25. Perform Data Cleaning, Data transformation using Python on any data set**

```python
import pandas as pd
import numpy as np

# ------------------ 1) Load Dataset ------------------
df = pd.read_excel("/mnt/data/Age Table.xlsx")
print("Original Dataset:")
print(df.head())


# ------------------ 2) DATA CLEANING ------------------

# (A) Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# (B) Fill missing numeric columns with mean
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())

# (C) Fill missing categorical columns with mode
cat_cols = df.select_dtypes(include=["object"]).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nAfter Filling Missing Values:")
print(df.isnull().sum())


# (D) Remove duplicate rows (if any)
df = df.drop_duplicates()
print("\nAfter Removing Duplicates:")
print(df)


# ------------------ 3) DATA TRANSFORMATION ------------------

# (A) Convert categorical to category-type
if "Age" in df.columns:
    df["Age"] = df["Age"].astype("category")

# (B) One-hot encoding (dummy variables)
df_dummies = pd.get_dummies(df, drop_first=True)
print("\nAfter One-Hot Encoding:")
print(df_dummies.head())

# (C) Normalize numeric columns (Min-Max scaling)
for col in num_cols:
    df_dummies[col+"_scaled"] = (df_dummies[col] - df_dummies[col].min()) / (df_dummies[col].max() - df_dummies[col].min())

print("\nAfter Scaling Numeric Columns:")
print(df_dummies.head())


# ------------------ 4) Save Cleaned + Transformed Data ------------------
df_dummies.to_csv("/mnt/data/Cleaned_Transformed_Data.csv", index=False)
print("\nCleaned & Transformed Dataset Saved.")
```
